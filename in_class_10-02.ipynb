XML demos in class

```python
import xml.etree.ElementTree as ET
```

```python
from pathlib import Path

EAD_relpath = Path("book-sample.xml")
```

```python
print(EAD_relpath)
```

```python
# parse XML = load + process

tree = ET.parse(EAD_relpath)
root = tree.getroot()
```

```python
print(root)
```

```python
print(ET.tostring(root))
```

```python
for element in root:
    print(element)
```

```python
for element in root:
    print(element.tag, "---", element.attrib, "---", element.text)
```

```python
for element in root.iter():
    print(element.tag, "---", element.attrib, "---", element.text)
```

```python
for element in root.iter():
    print(element.tag, ":", element.text)
```

```python
book = tree.find("{http://www.nogood.com/Book}book")
print(book.tag, book.attrib, book.text)
```

```python
for element in book:
    print(element.tag, element.text)
```

```python
books = root.findall(".//{http://www.nogood.com/Book}book")
print(books)
```

```python
# Problem 3

import json

parameters = {"fo": "json"}

def write_json_to_file(jsonfile, filename):
    with open(filename, "w", encoding="utf-8") as f:
        f.write(json.dumps(jsonfile))


resources = ["cph.3f05183/", "fsa.8d24709/", "highsm.64003/"]
n = 0
for resource in resources:
    paramsresource = requests.get(
        baseURL + resourceEndpoint + resource, params=parameters
    )
    jsonresource = paramsresource.json()
    n += 1
    filename = "resource" + str(n) + ".json"
    write_json_to_file(jsonresource, filename)
```

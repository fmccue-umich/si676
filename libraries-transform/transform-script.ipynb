```python
import json
from pathlib import Path
import csv

data = Path("data")
out = Path("..", "collection-data-extraction")
```

```python
metadata_file_path = out / "item-metadata"
print(metadata_file_path)
```

```python
for file in metadata_file_path.glob("item-metadata-*.json"):
    print(file)
```

```python
metadata_file_list = list()
for file in metadata_file_path.glob("item-metadata-*.json"):
    metadata_file_list.append(file)

print(len(metadata_file_list))
```

```python
with open(metadata_file_list[0], "r", encoding = "utf-8") as item:
    item_data = json.load(item)
    for element in item_data.keys():
        print(element, ":", item_data[element])
```

```python
with open(metadata_file_list[0], "r", encoding="utf-8") as item:
    item_data = json.load(item)
    try:
        item_title = item_data["title"]
    except:
        item_title = "No title found"
    try:
        item_id = item_data["id"]
    except:
        item_id = "No id found"
    try:
        item_date = item_data["date"]
    except:
        item_date = "No date found"
    try:
        item_contributors = item_data["contributor_names"][0]
    except:
        item_contributors = "No contributors found"

    row_dict = dict()

    row_dict["title"] = item_title
    row_dict["id"] = item_id
    row_dict["date"] = item_date
    row_dict["contributors"] = item_contributors
    print(row_dict)
```

```python
collection_info_csv = out / "collection_items_data.csv"
headers = ["title", "id", "date", "contributors"]

for item_file in metadata_file_list:
    with open(item_file, "r", encoding="utf-8") as item:
        item_data = json.load(item)
        try:
            item_title = item_data["title"]
        except:
            item_title = "No title found"
        try:
            item_id = item_data["id"]
        except:
            item_id = "No id found"
        try:
            item_date = item_data["date"]
        except:
            item_date = "No date found"
        try:
            item_contributors = item_data["contributor_names"][0]
        except:
            item_contributors = "No contributors found"

        row_dict = dict()

        row_dict["title"] = item_title
        row_dict["id"] = item_id
        row_dict["date"] = item_date
        row_dict["contributors"] = item_contributors

    row_dict = dict()

    row_dict["title"] = item_title
    row_dict["id"] = item_id
    row_dict["date"] = item_date
    row_dict["contributors"] = item_contributors

    with open(collection_info_csv, "a", encoding = "utf-8") as fout:
        writer = csv.DictWriter(fout, fieldnames = headers)
        writer.writerow(row_dict)
```

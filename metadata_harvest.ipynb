```python
import requests
import csv
```

```python
baseURL = "http://www.loc.gov/"
endpoint = "free-to-use/"
collection = "cats/"
parameters = {"fo" : "json"}
```

```python
collection_list = requests.get(baseURL + endpoint + collection, params=parameters)
```

```python
collection_list_items = collection_list.json()["content"]["components"][0]["masonry_gallery"][0]["items"]

for item in collection_list_items:
    print(item)
```

```python
outputs_csv = "cats_list.csv"
headers = ["image", "link", "title"]

with open(outputs_csv, "w", encoding="utf-8", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=headers)
    writer.writeheader()
    for item in collection_list.json()["content"]["components"][0]["masonry_gallery"][0]["items"]:
        item["title"] = item["title"].strip()
        writer.writerow(item)
```
